# Gözlem ve Değişken Seçimi

Pandas kullanımında en çok hata alınan konuları ele alacağız. **Loc ve ILoc**

```python
import numpy as np
import pandas as pd

m = np.random.randint(1,30,size=(10,3)) # numpy ile verisetimizi oluşturduk
df = pd.DataFrame(m, columns = ["var1","var2","var3"]) # pandas ile verisetimizi görüntüledik

df
```

**loc** : Tanımlandığı şekli ile seçim yapmak için kullanılır

* loc → Label Location (Etiket tabanlı)
* Sayısal indeks pozisyonları kullanır (0, 1, 2, 3...)
* Her zaman çalışır
* Boolean değer kabul eder

```python
df.loc[0:3] # 0'dan 3'e kadar, yani 3 dahil olmamak üzere almasını bekleriz ama loc'da alınır, ilocta alınmaz
```

**iloc** : Alışık olduğumuz indexleme mantığıyla seçim yapmak için kullanılır

* iloc → Integer Location (Pozisyon tabanlı)
* Gerçek indeks etiketlerini kullanır
* İndeks etiketleri ne ise onları arar
* Boolean değer kabul etmez, yalnızca int (df.iloc[df.var1>15, :2] bu hata verir çünkü df.var1>15 ifadesi true false)

```python
df.iloc[0:3] # bu normal, 0'dan 3'e kadar(3 dahil değil) şeklinde alıyor. ilactır ilac
```

```python
df.iloc[:3,:2] # 3.satırdaki değere ve 2.sütundaki değere kadar al(son değerler dahil değil)
```

```python
df.loc[:3,:"var2"]

# buradaki column kısmına iloctaki gibi :2 yazamadık çünkü yazsaydık hata alırdık. 
# Sebebi: loc: labeled Location tabanlı. eğer sütun kısmına :2 yazsaydık gidip sütun isimlerinde "2" arayacaktı
# iloc ise 2.indexte bulunan şeyi gider alır, bu index olarak değil, 2 isminde bişe var mı diye bakar
```

# DataFrame Birleştirme (Join) İşlemleri

Elimizde çoğu zaman 1 den fazla DataFrame olacak. Bu dataframelerin taşıdıkları bilgiler bir araya getirilmek istendiğinde DataFrame joint etme işlemleri yapılır.

```python
import numpy as np
import pandas as pd

dataset1 = np.random.randint(1, 30, size = (5,3))
df1 = pd.DataFrame(dataset1, columns=["var1","var2","var3"])
 
df1
```

```python
dataset2 = np.random.randint(1,30,size=(5,3))
df2 = pd.DataFrame(dataset2, columns=["var1","var2","var3"])
df2
```

```python
# Yol 1: Concat

pd.concat([df1,df2])

# iki adet dataframe'i birleştirmenin yollarından biri de concat kullanmaktır
# şu an iki adet dataframe'i birleştirdik
# buradaki tek problem indexler kendini tekrar ediyor. 0 1 2 3 4 sonra yine 0 1 2 3 4
```

```python
# tekrar eden index problemini gidermek için:

pd.concat([df1,df2],ignore_index=True) # ignore_index=True yaparsak indexleri birbirinin devamı yapar
```

Şu konulara **dikkat!!!!!!** edilmeli: Eğer her iki DataFrame'de de aynı sayıda ve aynı column isimleri yoksa bu sorun çıkartacaktır. Örneğin bir DataFrame'de var1 var2 varken diğerinde yalnızca var3 varsa bu sorundur. Ya da isim farklılıkları varsa da sorun olacaktır. Her ikisinde de 3 adet column olabilir ancak bunların isimleri de birebir aynı olmalı ki sorun çıkmasın. Bu sorundan kaçınmak, sadece kesişimleri olan kısımları eklemek için şu kodu kullanırız: 

**pd.concat([df1,df2], join="inner")**

**pd.concat([df1, df2.reindex(columns=df1.columns)]) dersek de df2'ye göre yapmış oluruz**

```python
# Şimdi bu bahsetmiş olduğum sorunu yaşamak için df2'nin column isimlerini değiştirelim ve neler olacak görelim!

df2.columns = ["var1","var2","deg1"] # var1 var2 var3 olmalıydı ama sonuncuyu değiştirdik
```

```python
df1
```

```python
df2 # Görüleceği üzere elimizde 2 tane farklı table isimli DataFrameler var.
```

```python
pd.concat([df1,df2], ignore_index=True) # deg1 değeri, df1 için tamamen anlamsız
```

```python
pd.concat([df1,df2],join="inner", ignore_index=True) 


# Her iki DataFrame'in kesişimlerine göre birleştirildi. Ortak sütunlar alınır yani. 
# Her ikisinde de olan var1 ve var2 tablosu var, o sebeple df1'de ki table 3 diğerinde, 
# df2'de ki dg1 de diğerinde olmadığından bu ikisi kaldırıldı
```

```python
pd.concat([df1, df2.reindex(columns=df1.columns)], ignore_index=True)

# burada da df1'de ki tüm tabloları getir, df2'de olupta 1 de olmayanı getirme demiş olduk.
# yani df1'e göre concat et
```

# İleri Seviye Toplulaştırma İşlemleri (Aggregate, Filter, Transform, Apply)

```python
import pandas as pd

df = pd.DataFrame({"gruplar":["A","B","C","A","B","C"],
                   "degisken1":[10,23,33,22,11,99],
                   "degisken2":[100,253,333,262,111,969]})
df
```

## Aggregate()

```python
df.groupby("gruplar").mean() # Burada önemli olan şey aggrigator(mean, median, std, etc) bir fonksiyon yazınca gruplama görünür hale geliyor
```

```python
df.groupby("gruplar").aggregate(["min", "max", "median"]) 

# aggregate fonksiyonu ile tek satırda birden fazla toplulaştırma işlemi yapabiliriz
# biz burada python'ın kendisine default tanımlı min,max ve median fonksiyonlarını kullandık
```

```python
df.groupby("gruplar").aggregate({"degisken1":"min", 
                                 "degisken2":"max"})

# her bir değişken(sütun) için farklı farklı istatistikler girebiliriz. Öncekilerde hep farklı değişkenler için aynı hesaplamaların sonucu
# gösterilirdi ancak şimdi aşağıda gördüğümüz tabloda değişken1 sütununda minimum değerler hesaplanırken değişken2 sütununda max değerler hesaplanıyor
```

## Filter()

Önceden basit anlamda filter yaptık. Şu anda da benzer şekilde dataframe içerisindeki değerler içerisinden filtrelediğimiz değerleri almaya çalışıcaz ancak kendi yazacağımız fonksiyona göre. Önceden pd.df(df.value1>15) gibi bir şey yapmıştık, şimdi bu parantez içerisine kendi yazdığımız fonksiyonu vereceğiz

```python
df = pd.DataFrame({"gruplar":["A","B","C","A","B","C"],
                   "degisken1":[10,23,33,22,11,99],
                   "degisken2":[100,253,333,262,111,969]})
df
```

```python
def filter_function(x): # A,B ve C değerlerinden degisken1 sütununda standart sapması 9'dan büyük olan değerleri getir
    return x["degisken1"].std() > 9
```

```python
df.groupby("gruplar").filter(filter_function)

# görüldüğü üzere standart sapması 9'dan küçük olan A ve B sınıflarını eledik, bu koşulu sağlayan C klasını aldık kendi yazdığımız fonksiyonla
```

# Pivot Tablolar

Verisetleri üzerinde bazı satır ve sütun işlemleri yaparak, verisetini amaca uygun getirmek için kullanılan yapılardır. Groupby işlemiyle karıştırılabilir, groupby'ın çok boyutlu halidir gibi düşünülebilir

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")

titanic.head()
```

```python
titanic.groupby("sex")["age"].mean()  

# kadınların yaş ortalamasıyla erkeklerin yaş ortalamasını aldık
# burada aslında ilkel bir pivot işlemi yapmış olduk. Titankteki insanların cinsiyetlerin yaş ortalamsına bakmış olduk
```

```python
titanic.groupby("sex")[["age"]].mean()

# burada sütun ekleyecek bu dediği için(2 adet köşeli parantez var "age"de) daha güzel görüntü elde ettik. aslında öncekinin aynısı
```

```python
titanic.groupby(["sex","class"], observed=True)[["survived"]].aggregate("mean").unstack()

# titanikteki kadın ve erkeklerin satın aldıkları bilete göre gemi battıktan sonra hayatta kalıp kalamama durumu
# aslında burada pivot görüntüsü elde etmiş olduk
```

```python
# Pivot ile Table

titanic.pivot_table("survived", index="sex", columns="class", aggfunc="mean")

# bize odaklandığın özellik nedir, indexin nedir, başka değerler var mı, hangi işlemi yapacaksın? 
# sorularını sorar ve onları tektek cevapladıktan sonra karşımıza pivot tablomuz gelmiş olur. aggfunc'ı yazmasaydık da mean yapardı çünkü default mean
```
