# Basit Doğrusal Regresyon (Linear Regression)

Temel Amaç: Bağımlı ve bağımsız değişkenler arasındaki ilişkiyi ifade eden doğrusal fonksiyonu bulmaktır.

Nasıl bulunur: Hata kareler toplamını minimize edecek katsayı tahminlerini bulmaya çalışarak yapılıyor



* b0 ve b1: Elimizde yoklar
* e: Hata
* x: Bağımsız değişkenler
* Y: Bağımlı değişken

Ana amaç y'yi tahmin etmek ancak öncesinde deneye deneye katsayı(b) değerlerini bulmaya/tahmin etmeye çalışıyoruz.



SSE: Hata kareler toplamıdır. Türev işlemleri yapılarak bulunur.
