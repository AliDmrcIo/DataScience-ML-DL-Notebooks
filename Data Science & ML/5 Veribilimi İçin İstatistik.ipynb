# Güven Aralığı

Anakütle parametresinin tahmini değerini kapsayabilecek iki sayıdan oluşan bir aralık bulunmasıdır. Yapmış olduğumuz herhangi bir ölçümün hassasiyetinin göstergesidir. Yani yapmış olduğumuz tahminlerin ne kadar güvenilir olduğuyla ilgili değer sunar bizlere.

Örneğin Örnek Teorisinde, örneklemden popülasyonun yaş ortalamasını bilmiştik. Güven Aralığını kullanırsak, "bana bir aralık ver, kesin o aralıkta olsun" gibi bir şey demiş ve bunu yüzdelerle ifade etmiş oluyoruz.

Örneğin web sitemizde kullanıcıların geçirdikleri sürenin ortalaması 180 saniye, standart sapması ise 40 saniye. Biz güven aralığı ile şunu diyoruz: "İstatistiksel olarak websitemizde kullanıcıların geçirdikleri süre, **%95 doğruluk oranı** ile 172-188 saniye **arasındadır**"

Güven Aralığını el yordamıyla bulma:
* Adım 1: n, ortalama, standart sapmayı bul
* Adım 2: Güven Aralığına karar ver. (95 mi 99 mu. bunu kendimiz belirliyoruz. genelde 95 seçilir)
  Z Tablo Değerini hesapla(1.96-2.57)
* Güven Aralığını hesapla:
  
  

n(örneklem sayısı)=100
ortalama=180
Standart Sapma=40





### Güven Aralığı Kod Uygulaması: Gerçek Hayattan Uyarlama

```python
import numpy as np
fiyatlar = np.random.randint(10,110,1000) # anket şirketi 1000 kişiyle görüştü ve 10 ile 110 arasında farklı fiyatlar aldı
```

```python
# 1.Yöntem: Güven Aralığı Manüel Hesaplama

ortalama = fiyatlar.mean()
standart_sapma = fiyatlar.std()
n = 1000
z = 1.96

guven_araligi_pozitif = ortalama + z*(standart_sapma)/(np.sqrt(n))
guven_araligi_negatif = ortalama - z*(standart_sapma)/(np.sqrt(n))

print(f"Kullanıcılar, {guven_araligi_negatif} TL ile {guven_araligi_pozitif} TL arasında ücret öderler")
```

```python
# 2.Yöntem: Güven Aralığı statsmodels Kütüphanesi Aracılığıyla Hesaplama

import statsmodels.stats.api as sms
sms.DescrStatsW(fiyatlar).tconfint_mean()

# Hocanın çalıştığı şirkette 3-4 ay boyunca böyle bir proje üzerinde onlarca kişi çalışmış. Çok önemli yani
```

# Olasılık

Olasılık: Olayların olabilirliği ya da olmayabilirliğinin sayısal ifadesidir. Bir olayın meydana gelmek için gerekli durum sayısının, tüm durumlar sayısına oranıdır.
- Belirsizlik durumunda kullandığımız matematik konularından biridir. Çıkarımsal istatistik ve veri bilimi için istatistik kapsamında bazı belirsiz durumlarını sayılarla belirtmek için kullanılır.

Rassal Değişken: Değerlerini bir deneyin sonuçlarından alan değişkenlere rassal değişkenler denir. Olasılık dağılımlarını kullanarak bazı olaylara ilişkin olasılıklar hesaplamak istediğimizde rassal değişkenleri kullanırız. 

Dağılım nedir: Evrende bazı olaylar gerçekleşir, bu olaylar ya da durumların sayısal karşılıklarının ortaya çıkardığı yapıya "Dağılım" denir.

Olasılık Fonksiyonu: Bir değişkenin herhangi bir değeri alması olasılığını hesaplamaya yarayan fonksiyondur

Kesikli ve Sürekli Olasılık: Elimizdeki değişkenin tipine göre hesaplama yapacak olduğumuz fonksiyon değişecektir. Dağılımlar değişecektir.

Kesikli Olasılık Dağılımları: Başarılı/Başarısız, Olumlu/Olumsuz gibi iki sonuçlu olaylarla ilgilendiğimizde kesikli olasılık kullanırız
- Bernoulli
- Binom
- Poisson

Sürekli Olasılık Dağılımları:
- Normal Dağılım
- Üniform Dağılım
- Üstel Dağılım

# Bernoulli Dağılımı



* p: Olasılığı ifade eder. Aradığımız durumun gerçekleşme olasılığı
* x: Kesikli değişkenimizin alacağı değer. bernoulli kapsamında 0 ya da 1 olabilir. örneğin yazı=1, tura=0
* E(X): Beklenen Değer
* Var(X): Varyans. İlgili olayın gerçekleşme olasılığı * gerçekleşmeme olasılığı

```python
from scipy.stats import bernoulli # olasılık değerlerini hesaplamak için scipy kütüpünü kullanıyoruz
```

```python
# Yazı Tura Örneği:

p = 0.6 # tura gelme olasılığı

rv = bernoulli(p)

rv.pmf(k=1) 

# probability mass functions'ın kısaltmasıdır pmf. k=1 diyince, ilgilendiğimiz olayın olasılığı true'dur. 
# bize sonucu verdi. 0.6 olasılıkla gerçekleşebilir true gelme olayı dedi
```

# Büyük Sayılar Yasası

Bir rassal değişkenin **uzun vadeli** kararlılığını tanımlayan olasılık teoremidir.

Örneğin bir madeni parayı 5 kere attık ve 4 kere tura, 1 kere yazı geldi. o zaman tura gelme olasılığı %80, yazı ise %20. E hani %50 idi? İşte Büyük sayılar yasası da, madeni parayı ne kadar **çok kez** atarsan o kadar %50'ye %50 oranına yakalaşırsın diyor

```python
import numpy as np

rng = np.random.RandomState(123)

for i in np.arange(1,21):
    deney_sayisi = 2**i
    yazi_turalar = rng.randint(0, 2, size = deney_sayisi)
    yazi_olasiliklari = np.mean(yazi_turalar)
    print("Atış Sayısı: ", deney_sayisi, "---", 'Yazı olasılığı: %.2f' % (yazi_olasiliklari*100))
```

# Hipotez Testi

Bir inanışı(savı, tahmini vb) test etmek için kullanılan istatistiksel bir tekniktir.



## Hipotezler ve Türleri

Yine, örneklemimizdeki şehrin yaş ortalaması 50'dir ifadesi için:



**Sıfır Hipotezi** : Belirli değerler verilerek kurulan hipotezlere H0/Null/Sıfır Hipotezi denir. 

**Alternatif Hipotez** : Sıfır Hipotezi neyse o değildir. Örneğin H0=50 ise, H1, !=50, >50 ya da <50 olabilir

## Hata Tipleri

Örneğin biz daha önceden bir şehrin yaş ortalamasına 2.000 örneklem ile 39 demiştik. Bu hipotezdir ve doğru da olabilir yanlış da. Her iki koşulda da biz bu hipotezi kabul de edebiliriz red de. Bu durumları gösteren bir Hata Tipleri görseli ve anlamları: 



1. satır, 1.Sütun (H0 doğru, H0 reddedilmedi) : Doğru bir şeyi kabul etmenin(reddetmemenin) adı **"Güven Düzeyi"**'dir
1. satır, 2.Sütun (H0 doğru, H0 reddedildi) : Doğru bir şeyi reddedmenin adı **"1.Tip Hata"**'dır
2. satır, 1.Sütun (H0 yanlış, H0 reddedilmedi) : Yanlış bir şeyi kabul etmenin(reddetmemenin) adı **"2.Tip Hata"**'dır
2. satır, 2.Sütun (H0 yanlış, H0 reddedildi) : Yanlış bir şeyi reddedmenin adı **"Testin Gücü"**'dür

## P-Value

Hipotezleri test etmek için vardır. Hipotezin P-Value'sü 0.05'den küçükse, H0 hipotezini reddederiz.

a = 0.05

**p < 0.05** olmalı.

P < 0.05 ise:
* H₀ hipotezi reddedilir
* H₁ (alternatif hipotez) desteklenir
* Ama bu H₁'in %100 doğru olduğu anlamına gelmez!

P ≥ 0.05 ise:
* H₀ hipotezi reddedilemez
* H₀'ın doğru olduğunu kanıtlamış olmayız, sadece reddetmek için yeterli kanıt yoktur



## Hipotez Testi Adımları

* Adım 1 : Hipotezlerin kurulması ve yönlerinin belirlenmesi(H0 ve H1 belirlenir)
* Adım 2 : Anlamlılık düzeyi ve tablo değerinin belirlenmesi(alfa : a)
* Adım 3 : Test istatistiğinin belirlenmesi ve hesaplanması
* Adım 4 : Hesaplanan test istatistiği ile alfaya karşılık gelen tablo değerinin karşılaştırılması. Test istatistiği (Zh) > Tablo değeri (Zt) ise H0 reddedilir
* Adım 5 : Sonuçlar Yorumlanır

Bazı kullanıcılara yeni entegre edilmiş olan ürün öneren ML modeli entegre edilmezken bazı kullanıcılara bu model entegre edilip kullanmaları sağlanıyor ve günün sonunda elde edilen gelirler kıyaslanıp attığımız taş ürküttüğümüz kurbağaya değdi mi diye bakılıyor. Hoca bu projede 7-8 ay çalışmış.

```python
# 1.Yol: Veri bize bu şekilde array olarak gelebilir

A = pd.DataFrame([30,27,21,27,29,30,20,20,27,32,35,22,24,23,25,27,23,27,23,
                 25,21,18,24,26,33,26,27,28,19,25]) # eski sistem

B = pd.DataFrame([37,39,31,31,34,38,30,36,29,28,38,28,37,37,30,32,31,31,27,
                 32,33,33,34,31,32,33,26,32,33,29]) # yeni sistem

A_B = pd.concat([A, B], axis = 1)
A_B.columns = ["A","B"]

A_B.head()
```

```python
# 2.Yol: Veri bize bu şekilde array olarak gelebilir: en boktan, en amele yol

A = pd.DataFrame([30,27,21,27,29,30,20,20,27,32,35,22,24,23,25,27,23,27,23,
                 25,21,18,24,26,33,26,27,28,19,25])

B = pd.DataFrame([37,39,31,31,34,38,30,36,29,28,38,28,37,37,30,32,31,31,27,
                 32,33,33,33,31,32,33,26,32,33,29])

#A ve A'nın grubu
GRUP_A = np.arange(len(A)) # A arrayinin boyutunda yeni bir array açtık
GRUP_A = pd.DataFrame(GRUP_A) # dataframe'e çevirdik
GRUP_A[:] = "A"  # içerisine ful A yazdık
A = pd.concat([A, GRUP_A], axis = 1) # hangi veri hangi gruba ait karışmasın diye 1.sütuna gelirleri, ikinci sütuna A harflerini koyduk

#B ve B'nin Grubu
GRUP_B = np.arange(len(B))
GRUP_B = pd.DataFrame(GRUP_B)
GRUP_B[:] = "B"
B = pd.concat([B, GRUP_B], axis = 1)

#Tüm veri
AB = pd.concat([A,B])
AB.columns = ["gelir","GRUP"]
print(AB.head())
print(AB.tail())
```

```python
oncesi = pd.DataFrame([123,119,119,116,123,123,121,120,117,118,121,121,123,119,
                       121,118,124,121,125,115,115,119,118,121,117,117,120,120,
                       121,117,118,117,123,118,124,121,115,118,125,115]) # eğitimden önce performanslar

sonrasi = pd.DataFrame([118,127,122,132,129,123,129,132,128,130,128,138,140,130,
                        134,134,124,140,134,129,129,138,134,124,122,126,133,127,
                        130,130,130,132,117,130,125,129,133,120,127,123]) # eğitimden sonra performanslar
```

```python
oncesi[:5]
```

```python
sonrasi[:5]
```

```python
#BIRINCI VERI SETI
AYRIK = pd.concat([oncesi, sonrasi], axis = 1)
AYRIK.columns = ["ONCESI","SONRASI"]
print("AYRIK Veri Seti: \n\n ", AYRIK.head(), "\n\n")

#IKINCI VERI SETI
#ONCESI FLAG/TAG'INI OLUSTURMA
GRUP_ONCESI = np.arange(len(oncesi))
GRUP_ONCESI = pd.DataFrame(GRUP_ONCESI)
GRUP_ONCESI[:] = "ONCESI"
#FLAG VE ONCESI DEGERLERINI BIR ARAYA GETIRME
A = pd.concat([oncesi, GRUP_ONCESI], axis = 1)
#SONRASI FLAG/TAG'INI OLUSTURMA
GRUP_SONRASI = np.arange(len(sonrasi))
GRUP_SONRASI = pd.DataFrame(GRUP_SONRASI)
GRUP_SONRASI[:] = "SONRASI"

#FLAG VE SONRASI DEGERLERINI BIR ARAYA GETIRME
B = pd.concat([sonrasi, GRUP_SONRASI], axis = 1)

#TUM VERIYI BIR ARAYA GETIRME
BIRLIKTE = pd.concat([A,B])
BIRLIKTE

#ISIMLENDIRME
BIRLIKTE.columns = ["PERFORMANS","ONCESI_SONRASI"]
print("'BIRLIKTE' Veri Seti: \n\n", BIRLIKTE.head(), "\n")
```

# Korelasyon Analizi

Değişkenler arasındaki ilişki, bu ilişkinin yönü ve şiddeti ile ilgili bilgiler sağlayan istatistiksel bir yöntemdir. Doğru orantı, ters orantı mevzusu



Mantıklı çünkü mükemmel pozitif korelasyonun olduğu tablo zaten doğru orantı tablosu normalde, diğer tabloları da bu mantıkla inceleyince tüm taşlar yerine oturuyor.

-----------------------------------------------------------------------------
**Hipotez:**




* H0: Değişkenler arasında korelasyon yoktur.
* H1: Değişkenler arasında korelasyon vardır.

-----------------------------------------------------------------------------

**Pearson Korelasyon Katsayısı ve Test İstatistiği:**



Pearson Korelasyon Katsayısı:
* xi, yi : xi ve yi gibi ifadeler gözlem birimlerini ifade eder.
* n : gözlem sayısı
* r : korelasyon katsayısı

Test İstatistiği:
* r : korelasyon katsayısı
* n : gözlem sayısı

-----------------------------------------------------------------------------
**Varsayımlar:**
* Normallik Dağılımı

 Varsayım Sağlanıyorsa -> Pearson Korelasyon Katsayısı

 Varsayım Sağlanmıyorsa -> Spearman Korelasyon Katsayısı

## Parametrik Hipotez Testi

### Korelasyon Katsayısı

```python
df["tip"].corr(df["total_bill_without_tip"])

# ön tanımlı olarak 0.57 olarak pearson korelasyon katsayısını verdi. 
# Pearson korelasyon katsayısı, normalde değişkenler için normallik dağılımı sağlandığında hesaplanır. Sağlanmıyorsa spearman
```

```python
df["tip"].corr(df["total_bill_without_tip"], method="spearman")

# varsayımımız sağlanmadığı için spearman korelasyon katsayısını kullanarak korelasyonu hesapladık
# şimdi bu sayı anlamlı mı testini yapmak için hipotez testine doğru geçiyore
```

### Korelasyon Hipotez Testi

## Nonparametrik Hipotez Testi

Bizim örneğimizde **varsayımlar sağlanmadığı** için **Spearman** ya da **Kendall Tau** isimli **nonparametrik yöntemleri** kullanırız

```python
stats.spearmanr(df["tip"], df["total_bill_without_tip"]) # değişkenler arasında ilişki yoktur diyen H0 hipotezi reddedildi
```

```python
stats.kendalltau(df["tip"], df["total_bill_without_tip"]) # değişkenler arasında ilişki yoktur diyen H0 hipotezi reddedildi

# normalde spearman yeterli olacaktır ancak çok çok emin olmak için ben bir tane dahaya bakıyım dersen bunu kullan. 
# biraz daha temkinli bir korelasyon katsayısıdır
```
