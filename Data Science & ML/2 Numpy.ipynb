# Array Birleştirme(Concatenation)

```python
import numpy as np
```

```python
x = np.array([1,2,3])
y = np.array([4,5,6])

# şimdi bunları birleştiricez
```

```python
np.concatenate([x,y]) # burada x ve y arraylerinin birleşimini yaptık
```

```python
# iki boyut
```

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([[7,8,9],
              [10,11,12]])

np.concatenate([a,b])
```

```python
# eğer y ekseninde birleştirmek istersem axis değerini güncellerim. Axis eksen belirtmek için vardır. 1=x  0=y

np.concatenate([a,b],axis=1) # x eksenine ekleme yaptı, sütun bazında değil, satır bazında çoğaldı arrayimiz
```

# Array Ayırma(Splitting) 

```python
x = np.array([1,2,3,99,99,3,2,1]) 

# şimdi buradan değerleri sırasıyla 1,2,3 99,99 ve 3,2,1 olan 3 farklı array çıkartıcaz
```

```python
np.split(x, [3,5]) # x arrayinin 0-3, 3-5, 5-8 eleman arasını 3 farklı arraye böler. 

# Verilen index numaraları dahil değildir. Yani burada 0.indexten 3.index'e kadar(3 dahil değil), 3'ten 5'e kadar 

a,b,c = np.split(x, [3,5]) # burada da sırasıyla arrayleri değerlere atar
```

# Fancy Index ile Elemanlara Erişmek

İleride hem Pandas kütüphanelerinde hem de Numpy arraylerinde çok sıkça kullanacağız. Bu sebeple burası çok önemli!

```python
import numpy as np
v = np.arange(0,30,3)
v
```

Örneğin ben bu array'de 1, 3 ve 5 numaralı indexleri almak istersem şöyle yapabilirim:

```python
v[1]
```

```python
v[3]
```

```python
v[5]
```

```python
# ya da bir dizi içerisine koyup diziyi oynatabilirim

elemanListesi=np.array([v[1],v[3],v[5]])
elemanListesi
```

```python
# ancak elimizdeki index sayısı ve liste içerisindeki eleman sayısı arttıkça işler karışmaya başlaycak
```

# Numpy ile İki Bilinmeyenli Denklem Çözümü

5*x0 + x1 = 12

x0 + 3*x1 = 10

Bu denklemi bir tane kütüphane kullanarak çözdüreceğiz

Bunu yaparken bu ifadeyi numpyın anlayacağı bir şekle sokmak gerekiyor. Bunu yaparken de katsayıları bir diziye, diğerlerini bir diziye koyduktan sonra solve isimli fonksiyonu kullanarak bunu çözceceğiz

```python
a = np.array([[5,1],[1,3]]) # Katsayılar Arrayi. 1.denklemde 5 ve 1 katsayıları, 2.denklemde 3 ve 1 katsayıları var
b = np.array([12,10]) # eşitliğin sağındaki değerler
```

```python
a
```

```python
b
```

```python
x = np.linalg.solve(a,b) 

# lineercebir için olan linalg kütüpünün solve fonksiyonunu 2 bilinmeyenli denklem için kullandık
```
