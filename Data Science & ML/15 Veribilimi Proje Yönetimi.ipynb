# Projeye Başlarken Sorulacak Sorular



# Adım 1


Projeye yapılacak iyi bir başlangıç, projenin kalan gidişatını belirleyecektir. 

* Problem tanımı: "Bir problemi çözmek için 1 saatim varsa, 55 dakikasını problemi anlamaya, 5 dakikasını çözüme ayırırım" diyor Albert Einstein. Bu en en en kritik aşamadır. Problemi anlarsan çözüm de kafada canlanacaktır.
* Beklenti yönetimi: proje paydaşlarının(ilgili departmanlar, yöneticiler) hepsiyle tam anlamıyla detaylı fikir alışverişi yapılmalı, her birinin ayrı ayrı beklentileri öğrenilmeli ve buna göre projeye başlanmalıdır.
* Projemizi, ilgilerin ne olması durumunda başarılı bulacağı detaylıca öğrenilmeli ve netleştirilmelidir. Tam olarak proje nasıl olursa bu projeye başarılı dersiniz?

# Adım 4


Özellik mühendisliğinden genel olarak anlaşılması gereken şey şudur: Verisetimizde zengin bir bilgi sunacağını varsayarak hedefimize göre değişken üretmektir. Ürettiğimiz değişkenleri denetlemektir, sorasında ise dataseti tamamen modelleme öncesine hazır hale getirmektir.

* Tidy süreci: analitik herhangi bir çalışma öncesinde verinin yapısal anlamda düzenli hale getirilmesidir. Düzenli'den kasıt; her bir satırın ilgilenilen gözlem birimini, her bir sütunun ise bir değişkeni ifade etme durumundan emin olmaktır.
* Keşiftçi veri analizi: en çok mesai yaptığımız alan zaten.
* Feature engineering: Burada veri manipülasyonu kapsamında; eksik veri incelemesi, aykırı gözlem incelemesi, tutarsızlıkların incelenmesi gibi çok değişkenli istatistiksel yöntemlerin uygulanması söz konusudur.

# Adım 5


Bir önceki aşamada veriyi yapısal olarak bir araya getirip işlemeye hazır hale getirdikten sonra artık modellemeye hazır hale getirmekte sıra, buna da veri önişleme diyoruz. eksik veri incelemesi, aykırı gözlem incelemesi gibi işlemler yapılır, zaten detaylıca üzerinde durduk.

# Adım 6


Elimizde her türlü kirlilikten, dağınıklıktan, anormallikten, gürültüden arınmış bir veri var, sıra geldi bu hazır veriyi modele vermeye. Bu aşamada algoritmalar(modelleme yöntemleri) denenir ve model tuning işlemleri ve modellerin farkları incelendikten sonra mevcut veri için en uygun modele karar verilmesi ve işlemlerin başlanması gereklidir.
